# Hybrid plant scheduler: QAOA + classical incentives

Enter any fleet of plants (up to 16), and this schedules the day hour by hour:

1. **Classical incentives**: Lagrangian prices + each plant's cost-to-go, so each hour's QUBO knows what a startup or shutdown costs later in the day.
2. **QAOA** samples each hour's on/off choice from that QUBO (statevector simulator).
3. **Classical repair** fixes what's left and gives a certified bound on the gap, then everything is compared with the **MILP optimum** as the standard.

Kernel: **qBraid-SDK (v0.12.2)**. Qubits per hour = plants + 3 slack bits. Up to ~12 plants runs in seconds to a minute; 16 plants (19 qubits) takes several minutes per run.

Benchmark and analysis: `hybrid_uc/RESULTS.md`. Command-line version: `python -m hybrid_uc.run --help`.
